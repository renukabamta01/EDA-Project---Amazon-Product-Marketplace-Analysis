### 2. Hypothesis Testing
- Hypothesis testing helps us determine whether an observed difference or relationship in the sample is statistically significant or could reasonably have occurred by chance.
  
- Usually, we use a 5% significance level (α = 0.05).

#### 1. T-Test
- A T-test is used to compare the means of two groups.

```python
# Step 2: Python Code

from scipy.stats import ttest_ind

# Separate the two groups
best_seller = df[df['is_best_seller'] == True]['Current/discounted_price'].dropna()

non_best_seller = df[df['is_best_seller'] == False]['Current/discounted_price'].dropna()

# Independent T-Test
t_stat, p_value = ttest_ind(
    best_seller,
    non_best_seller,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

#### 2. Chi-Square Test
- Chi-Square is used when both variables are categorical.

```python
from scipy.stats import chi2_contingency

# Create contingency table
table = pd.crosstab(
    df['is_best_seller'],
    df['is_sponsored']
)

print(table)
```

```python
chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

#### 3. ANOVA
- ANOVA = Analysis of Variance.
- It is used to compare the means of three or more groups.

```python
df['rating_group'] = pd.cut(
    df['Rating'],
    bins=[0, 3, 4, 5],
    labels=['Low', 'Medium', 'High']
)
```

```python
from scipy.stats import f_oneway

low = df[
    df['rating_group'] == 'Low']['Current/discounted_price'].dropna()

medium = df[
    df['rating_group'] == 'Medium']['Current/discounted_price'].dropna()

high = df[
    df['rating_group'] == 'High']['Current/discounted_price'].dropna()

f_stat, p_value = f_oneway(
    low,
    medium,
    high
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

#### 4. Correlation Significance Test

```python
from scipy.stats import pearsonr

r, p_value = pearsonr(
    df['Rating'].dropna(),
    df['Number_of_reviews'].dropna()
)

print("Correlation:", r)
print("P-value:", p_value)
```

## 3. Confidence Interval Analysis
- A Confidence Interval (CI) gives us a range of values that is likely to contain the true population parameter, based on our sample.

```python
# 1. Confidence Interval for Average Rating

import numpy as np
from scipy import stats
```

```python
rating = df["Rating"].dropna()
```

```python
mean_rating = rating.mean()

confidence_level = 0.95
alpha = 1 - confidence_level

confidence_interval = stats.t.interval(
    confidence_level,
    df=len(rating)-1,
    loc=mean_rating,
    scale=stats.sem(rating)
)

print("Mean Rating:", mean_rating)
print("95% Confidence Interval:", confidence_interval)
```

```python
# 2. Confidence Interval for Product Price

price = df['Current/discounted_price'].dropna()

mean_price = price.mean()

confidence_interval = stats.t.interval(
    0.95,
    df=len(price)-1,
    loc=mean_price,
    scale=stats.sem(price)
)

print("Mean Price:", mean_price)
print("95% Confidence Interval:", confidence_interval)
```

```python
# Confidence Interval for Number of Reviews

reviews = df['Number_of_reviews'].dropna()

mean_reviews = reviews.mean()

confidence_interval = stats.t.interval(
    0.95,
    df=len(reviews)-1,
    loc=mean_reviews,
    scale=stats.sem(reviews)
)

print("Mean Reviews:", mean_reviews)
print("95% Confidence Interval:", confidence_interval)
```

```python
# Create a Reusable Function

def confidence_interval(data, confidence=0.95):
    
    data = data.dropna()
    
    mean = data.mean()
    
    ci = stats.t.interval(
        confidence,
        df=len(data)-1,
        loc=mean,
        scale=stats.sem(data)
    )
    
    return mean, ci
```

```python
mean_rating, ci_rating = confidence_interval(df['Rating'])

print("Mean Rating:", mean_rating)
print("95% CI:", ci_rating)
```

```python
mean_price, ci_price = confidence_interval(
    df['Current/discounted_price']
)

print("Mean Price:", mean_price)
print("95% CI:", ci_price)
```

### 4. Feature Engineering
- Feature Engineering means creating new meaningful columns from existing columns to make the data more useful for analysis and business decision-making.

```python
# 1. Discount Percentage

df['Discount_Percentage'] = (
    (df['Listed_price'] - df['Current/discounted_price'])
    / df['Listed_price']
) * 100
```

```python
df[['Listed_price','Current/discounted_price','Discount_Percentage']].head()
```

```python
# 2. Discount Category

df['Discount_Category'] = pd.cut(df['Discount_Percentage'],bins=[-1, 10, 30, 50, 100],
    labels=[
        'Low Discount',
        'Medium Discount',
        'High Discount',
        'Very High Discount'
    ]
)
```

```python
df['Discount_Category'].value_counts()
```

```python
# 3. Rating Category

df['Rating_Category'] = pd.cut(
    df['Rating'],
    bins=[0, 3, 4, 5],
    labels=[
        'Low Rating',
        'Medium Rating',
        'High Rating'
    ]
)
```

```python
df['Rating_Category'].value_counts()
```

```python
# 4. Review Engagement Category

df['Review_Engagement'] = pd.cut(
    df['Number_of_reviews'],
    bins=[-1, 100, 1000, 10000, float('inf')],
    labels=[
        'Low',
        'Medium',
        'High',
        'Very High'
    ]
)
```

```python
df['Review_Engagement'].value_counts()
```

```python
# 5. Price Category

df['Price_Category'] = pd.cut(
    df['Current/discounted_price'],
    bins=[-1, 20, 50, 100, float('inf')],
    labels=[
        'Budget',
        'Mid-Range',
        'Premium',
        'High-Premium'
    ]
)
```

```python
# 6. Promotion Flag

df['Promotion_Flag'] = (
    (df['is_couponed'] == True) |
    (df['is_sponsored'] == True)
)
```

```python
# 7. Bestseller + High Rating Feature

df['High_Rated_Bestseller'] = (
    (df['Rating'] >= 4) &
    (df['is_best_seller'] == True)
)
```

```python
# Price per Review

df['Price_per_Review'] = (
    df['Current/discounted_price']
    / df['Number_of_reviews'].replace(0, np.nan)
)
```

```python
# Product Performance Segment

df['Performance_Segment'] = np.where(
    (df['Rating'] >= 4) &
    (df['Number_of_reviews'] >= 1000),
    'Strong Performance',
    'Other'
)
```
