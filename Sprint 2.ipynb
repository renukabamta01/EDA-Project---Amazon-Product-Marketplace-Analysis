### 5. Group-Based Analysis

```python
# Bestseller Status → Average Purchases, Rating and Reviews

bestseller_analysis = df.groupby("is_best_seller").agg(
    avg_rating=("Rating", "mean"),
    avg_reviews=("Number_of_reviews", "mean"),
    avg_purchases=("bought_in_last_month", "mean"),
    avg_price=("Current/discounted_price", "mean")
)

bestseller_analysis
```

```python
# Coupon Status → Average Purchases and Price

coupon_analysis = df.groupby("Coupon_Status").agg(
    avg_rating=("Rating", "mean"),
    avg_reviews=("Number_of_reviews", "mean"),
    avg_purchases=("bought_in_last_month", "mean"),
    avg_price=("Current/discounted_price", "mean")
)

coupon_analysis
```

```python
# Sponsorship Status → Product Performance

sponsored_analysis = df.groupby("is_sponsored").agg(
    avg_rating=("Rating", "mean"),
    avg_reviews=("Number_of_reviews", "mean"),
    avg_purchases=("bought_in_last_month", "mean"),
    avg_price=("Current/discounted_price", "mean")
)

sponsored_analysis
```

```python
# Price Groups → Purchases and Ratings
# Step 1: Create price groups

df["Price_Group"] = pd.cut(
    df["Current/discounted_price"],
    bins=[0, 500, 1000, 2000, float("inf")],
    labels=["Low", "Medium", "High", "Premium"]
)
```

```python
# Step 2: Analyze each group

price_analysis = df.groupby("Price_Group", observed=True).agg(
    avg_purchases=("bought_in_last_month", "mean"),
    avg_rating=("Rating", "mean"),
    avg_reviews=("Number_of_reviews", "mean"),
    product_count=("Title", "count")
)

price_analysis
```

```python
# Promotion Combination → Average Purchases

df["Promotion_Group"] = (
    df["is_sponsored"].astype(str)
    + " + "
    + df["is_couponed"].astype(str)
)
```

```python
promotion_analysis = df.groupby("Promotion_Group").agg(
    avg_purchases=("bought_in_last_month", "mean"),
    avg_rating=("Rating", "mean"),
    avg_reviews=("Number_of_reviews", "mean"),
    avg_price=("Current/discounted_price", "mean"),
    product_count=("Title", "count")
)

promotion_analysis
```

```python
# Price Group vs Average Purchases

price_analysis["avg_purchases"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Purchases by Price Group")
plt.xlabel("Price Group")
plt.ylabel("Average Purchases in Last Month")
plt.show()
```

```python
# Bestseller vs Average Purchases

bestseller_analysis["avg_purchases"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Purchases by Bestseller Status")
plt.xlabel("Bestseller Status")
plt.ylabel("Average Purchases in Last Month")
plt.xticks(rotation=45)
plt.show()
```

```python
# 2. Bestseller Status — Count Plot

plt.figure(figsize=(7, 5))

sns.countplot(
    x="Bestseller_Status",
    data=df
)

plt.title("Product Distribution by Bestseller Status")
plt.xlabel("Bestseller Status")
plt.ylabel("Number of Products")
plt.show()
```

```python
# 3. Price Distribution — Box Plot

plt.figure(figsize=(8,5))

sns.boxplot(x=df["Current/discounted_price"])
plt.title("Distribution of Current/Discounted Price")
plt.xlabel("Current/Discounted Price")
plt.show()

# Shows median price, spread, and possible high-price outliers.
```

```python
# 4. Price vs Purchases — Scatter Plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="Current/discounted_price",
    y="bought_in_last_month",
    data=df
)

plt.title("Price vs Monthly Purchases")
plt.xlabel("Current/Discounted Price")
plt.ylabel("Bought in Last Month")
plt.show()

# Helps visualize whether product price and recent purchase activity show a visible relationship.
```

```python
# 5. Performance Distribution — Violin Plot

plt.figure(figsize=(8, 5))

sns.violinplot(
    x="Performance",
    y="bought_in_last_month",
    data=df,
    order=["Low", "Medium", "High", "Very High"]
)

plt.title("Purchase Distribution Across Performance Levels")
plt.xlabel("Performance Level")
plt.ylabel("Bought in Last Month")
plt.show()

# Shows the distribution of purchase activity across performance groups
```

```python
# 6. Promotion Analysis — Bar Chart

promotion_performance = df.groupby("is_sponsored")["bought_in_last_month"].mean().reset_index()

plt.figure(figsize=(7,5))
sns.barplot(x="is_sponsored",y="bought_in_last_month",data = promotion_performance)

plt.title("Average Purchases by sponsored Status")
plt.xlabel("Sponsored Status")
plt.ylabel;("Average Purchases")
plt.show()

# Makes differences in average purchase activity between sponsored groups easy to compare.
```

```python
# 7. Correlation Heatmap

numeric_cols = [
    "Rating",
    "Number_of_reviews",
    "bought_in_last_month",
    "Current/discounted_price",
    "Listed_price"
]

correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()

# This is particularly useful for Correlation & Relationship Analysis section.
```
