### Handling Missing Values

```python
df.isnull().sum()
```

#### For Numerical Column
- For Skewed Data(Right or Left Skewed)/Outliers ---> Use Median
- For Symmetric Data ---> Use Mean
#### Fro Categorical Column ---> USe Mode

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].mean())
```

```python
df["Number_of_reviews"] = df["Number_of_reviews"].fillna(df["Number_of_reviews"].median())
```

```python
df["bought_in_last_month"] = df["bought_in_last_month"].fillna(df["Number_of_reviews"].mode())
```

```python
df["Current/discounted_price"] = df["Current/discounted_price"].fillna(df["Current/discounted_price"].median())
```

```python
result = df["Listed_price"] == "No Discount"
df.loc[result,"Listed_price"] = df.loc[result,"Current/discounted_price"]
```

```python
df["Listed_price"].isna().sum()
```

```python
df["Listed_price"]
```

```python
df["Listed_price"] = df["Listed_price"].fillna(df["Listed_price"].median())
```

```python
import numpy as np

df["bought_in_last_month"] = (
    df["bought_in_last_month"]
    .str.extract(r'([\d.]+)\s*([Kk]?)')
    .assign(
        number=lambda x: pd.to_numeric(x[0], errors="coerce"),
        multiplier=lambda x: x[1].map({"K": 1000, "k": 1000}).fillna(1)
    )
    .eval("number * multiplier")
)
```

```python
df["bought_in_last_month"] = df["bought_in_last_month"].fillna(df["bought_in_last_month"].median())
```

```python
df["bought_in_last_month"].dtype
```

```python
df["buy_box_availability"] = df["buy_box_availability"].fillna(df["buy_box_availability"].mode()[0])
```

```python
df["delivery_details"] = df["delivery_details"].fillna("Not Available")
```

```python
df["sustainability_badges"] = df["sustainability_badges"].fillna("No Badge")
```

```python
df.drop("product_url", axis=1, inplace=True)
```

#### Outlier Analysis

```python
# Identify potential outliers in appropriate numerical features.
# Use suitable statistical and visualization techniques.
# Decide whether unusual observations should be:
# Retained
# Removed
# Transformed
# Every decision must be justified.

numerical_columns = ["Rating","Number_of_reviews","Current/discounted_price",
                "Listed_price","collected_at"]


for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) | (df[col] > upper)
    ]

    print(f"{col}: {len(outliers)} outliers")
```
